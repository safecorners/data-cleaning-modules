## 0. 데이터 분석 첫 5분 — 5가지 메서드

받은 데이터 처음 보면 이 순서로 살펴보세요:

```
1. shape       — 행/열 개수
2. head()      — 첫 몇 줄 (모양 확인)
3. info()      — 컬럼 타입, 결측 여부
4. describe()  — 수치 컬럼 요약 통계
5. value_counts() — 카테고리 컬럼 빈도
```

이 다섯 가지가 **데이터 분석의 첫 5분**이에요.


## 1. `shape` — 행/열 개수

가장 먼저 확인할 것.


````python
import pandas as pd

graduate = pd.read_csv('data/한국교육개발원 시도 시군구별 졸업자 진학자 진학률_20240401.csv', encoding='cp949')
print(f"행: {graduate.shape[0]}, 열: {graduate.shape[1]}")
print(f"shape: {graduate.shape}")
````

## 2. `head()` — 첫 몇 줄


````python
# 기본 5줄
graduate.head()
````

````python
# 더 많이 보고 싶으면 인자
graduate.head(10)
````

## 3. `info()` — 컬럼 타입 + 결측 여부

가장 유용한 메서드 중 하나. 한 번에 많은 정보가 나옴.


````python
graduate.info()
````

**info()가 알려주는 것**:
- 각 컬럼의 데이터 타입 (`int64`, `float64`, `object`)
- 각 컬럼의 Non-Null 개수 → **결측 개수 추정 가능** (전체 - non-null)
- 메모리 사용량

`object`는 보통 **문자열** 컬럼이에요.


## 4. `describe()` — 수치 컬럼 요약 통계


````python
graduate.describe()
````

**describe()가 알려주는 것**:
- `count` — 결측 아닌 값 개수
- `mean`, `std` — 평균, 표준편차
- `min`, `max` — 최솟값, 최댓값
- `25%`, `50%`, `75%` — 분위수 (중앙값 = 50%)

**용도**: 이상치 빠르게 발견 (`max`가 비상식적으로 크면 의심)


## 5. `value_counts()` — 카테고리 컬럼 빈도


````python
# 어떤 연도들이 있나?
graduate['연도'].value_counts()
````

````python
# 어떤 시도들이 있나?
graduate['시도'].value_counts()
````

**용도**:
- 카테고리 종류와 개수 한눈에
- 카테고리 표기 일관성 체크 (`'서울'`과 `'서울특별시'`가 섞여 있으면 발견)


## 6. `hist()` — 분포 시각화


````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.hist(graduate['졸업자'], bins=30, edgecolor='black')
plt.xlabel('Graduates')
plt.ylabel('Frequency')
plt.title('Distribution of graduates per district')
plt.grid(alpha=0.3)
plt.show()
````

````python
# 전체 수치 컬럼 한 번에 (편함!)
graduate.hist(figsize=(10, 6), bins=20)
plt.tight_layout()
plt.show()
````

## 8. ⚠️ 함정 / 주의사항

### 8.1 `describe()`는 수치 컬럼만 봄
문자열 컬럼은 기본 describe()에서 빠짐.  
**해결**: `df.describe(include='all')`로 전체 보기.

### 8.2 `info()`로 결측 알아채기
"Non-Null Count"가 전체 행 수보다 작으면 결측 있음.  
명시적 확인은 `df.isnull().sum()` (M02 모듈).
